
# 🔍 Rabin-Karp Algorithm

The **Rabin-Karp algorithm** is a string searching algorithm that uses hashing to find any one of a set of pattern strings in a text. It is especially useful when checking for multiple pattern matches.

---

## 🧠 How It Works

- Compute the hash of the pattern and the hash of each window of the text.
- Slide the window over the text and compare the hash values.
- If the hash matches, verify the actual characters.

---

## 🧮 Hashing

We use a rolling hash function to compute new hashes from previous ones efficiently.

**Formula:**
```
hash(s) = (s[0]*d^(m-1) + s[1]*d^(m-2) + ... + s[m-1]) % q
```
Where:
- `d` is the number of characters in the input alphabet (e.g., 256 for extended ASCII)
- `q` is a prime number to reduce collisions


````

def rabin_karp(pattern, text, d=256, q=101):
    M = len(pattern)
    N = len(text)
    p_hash = 0    # Hash value for pattern
    t_hash = 0    # Hash value for text
    h = 1
    result = []

    # The value of h would be "pow(d, M-1)%q"
    for i in range(M - 1):
        h = (h * d) % q

    # Calculate the hash value of pattern and first window of text
    for i in range(M):
        p_hash = (d * p_hash + ord(pattern[i])) % q
        t_hash = (d * t_hash + ord(text[i])) % q

    # Slide the pattern over text
    for i in range(N - M + 1):
        # If hash values match then only check characters one by one
        if p_hash == t_hash:
            if text[i:i + M] == pattern:
                result.append(i)

        # Calculate hash value for next window of text
        if i < N - M:
            t_hash = (d * (t_hash - ord(text[i]) * h) + ord(text[i + M])) % q
            # Convert negative value of t_hash to positive
            if t_hash < 0:
                t_hash = t_hash + q

    return result
````

````

# Example usage
text = "ABCCDDAEFG"
pattern = "CDD"
positions = rabin_karp(pattern, text)
print("Pattern found at positions:", positions)
````
