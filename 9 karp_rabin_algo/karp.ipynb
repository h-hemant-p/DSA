# Karp-Rabin Algorithm 
## Overview
- **Input**: A text string and a pattern string.
- **Output**: List of starting indices where the pattern occurs in the text.
- **Method**: Uses a rolling hash to compare the pattern with substrings of the text, verifying matches to handle hash collisions.

## Parameters
- `d`: Number of possible characters (e.g., 256 for ASCII).
- `q`: A prime number to reduce hash collisions (e.g., 101).


```python
def karp_rabin(text, pattern):
    # Input validation
    if not text or not pattern or len(pattern) > len(text):
        return []

    # Initialize variables
    d = 256  # Number of characters in input alphabet (ASCII)
    q = 101  # A prime number for modulo
    m = len(pattern)  # Length of pattern
    n = len(text)  # Length of text
    matches = []  # Store indices of matches
    h = 1  # Used for rolling hash calculation

    # Calculate h = d^(m-1) % q
    for i in range(m - 1):
        h = (h * d) % q

    # Calculate initial hash values for pattern and first window of text
    pattern_hash = 0
    window_hash = 0
    for i in range(m):
        pattern_hash = (d * pattern_hash + ord(pattern[i])) % q
        window_hash = (d * window_hash + ord(text[i])) % q

    # Slide pattern over text
    for i in range(n - m + 1):
        # Check if hash values match
        if pattern_hash == window_hash:
            # Verify character by character
            match = True
            for j in range(m):
                if pattern[j] != text[i + j]:
                    match = False
                    break
            if match:
                matches.append(i)

        # Calculate hash for next window
        if i < n - m:
            window_hash = (d * (window_hash - ord(text[i]) * h) + ord(text[i + m])) % q
            # Handle negative hash values
            if window_hash < 0:
                window_hash += q

    return matches
```

## Example Usage
Let's test the algorithm with a sample text and pattern.

```python
# Test the Karp-Rabin algorithm
text = "AABAACAADAABAAABAA"
pattern = "AABA"
result = karp_rabin(text, pattern)
print(f"Text: {text}")
print(f"Pattern: {pattern}")
print(f"Pattern found at indices: {result}")
```

## Explanation of Code
- **Hash Calculation**: The hash for the pattern and initial text window is computed using a polynomial rolling hash.
- **Rolling Hash**: For each subsequent window, the hash is updated by removing the leading character's contribution and adding the next character's contribution.
- **Collision Handling**: If hash values match, a character-by-character check confirms the match to avoid false positives.
- **Output**: Returns a list of starting indices where the pattern is found.

## Notes
- The choice of `q` (prime number) and `d` (alphabet size) affects the likelihood of hash collisions.
- The algorithm is efficient for large texts and multiple patterns but may degrade to O(n·m) in the worst case with many collisions.
