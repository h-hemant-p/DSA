## 1. Introduction to Binary Search

Binary search works on **sorted arrays** by repeatedly dividing the search interval in half. It compares the target value to the middle element of the array:
- If the target equals the middle element, the search is complete.
- If the target is less, it searches the left half.
- If the target is greater, it searches the right half.

**Prerequisites**:
- The array must be sorted.
- The array should support random access (via indexing).

**Time Complexity**: O(log n)  
**Space Complexity**: O(1) for iterative, O(log n) for recursive (due to call stack).

## 2. Iterative Binary Search

The iterative approach uses a loop to find the target element.

```python
def binary_search_iterative(arr, target):
    left, right = 0, len(arr) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        # Check if target is at mid
        if arr[mid] == target:
            return mid
        # If target is less, ignore right half
        elif arr[mid] > target:
            right = mid - 1
        # If target is greater, ignore left half
        else:
            left = mid + 1
    
    return -1  # Target not found

# Example usage
arr = [2, 3, 4, 10, 40, 50, 60, 70]
target = 10
result = binary_search_iterative(arr, target)
print(f"Iterative Binary Search: Element {target} found at index {result}")  # Output: Element 10 found at index 3

target = 55
result = binary_search_iterative(arr, target)
print(f"Iterative Binary Search: Element {target} found at index {result}")  # Output: Element 55 found at index -1
```

## 3. Recursive Binary Search

The recursive approach divides the problem into smaller subproblems.

```python
def binary_search_recursive(arr, target, left, right):
    if left > right:
        return -1
    
    mid = (left + right) // 2
    
    # Check if target is at mid
    if arr[mid] == target:
        return mid
    # If target is less, search left half
    elif arr[mid] > target:
        return binary_search_recursive(arr, target, left, mid - 1)
    # If target is greater, search right half
    else:
        return binary_search_recursive(arr, target, mid + 1, right)

# Wrapper function for easier usage
def binary_search(arr, target):
    return binary_search_recursive(arr, target, 0, len(arr) - 1)

# Example usage
arr = [2, 3, 4, 10, 40, 50, 60, 70]
target = 10
result = binary_search(arr, target)
print(f"Recursive Binary Search: Element {target} found at index {result}")  # Output: Element 10 found at index 3

target = 55
result = binary_search(arr, target)
print(f"Recursive Binary Search: Element {target} found at index {result}")  # Output: Element 55 found at index -1
```

### 4.2 Using NumPy's searchsorted

```python
# Using np.searchsorted
arr = np.array([2, 3, 4, 10, 40, 50, 60, 70])
target = 10
index = np.searchsorted(arr, target)
print(f"NumPy searchsorted: Element {target} found at index {index}")  # Output: Element 10 found at index 3

# Verify if element exists
if index < len(arr) and arr[index] == target:
    print(f"Element {target} is present at index {index}")
else:
    print(f"Element {target} not found")  # Won't trigger for target=10

target = 55
index = np.searchsorted(arr, target)
print(f"NumPy searchsorted: Element {target} found at index {index}")
if index < len(arr) and arr[index] == target:
    print(f"Element {target} is present at index {index}")
else:
    print(f"Element {target} not found")  # Output: Element 55 not found
```

### 4.3 Custom Binary Search with NumPy Arrays

You can adapt the iterative binary search for NumPy arrays.

```python
def binary_search_numpy(arr, target):
    left, right = 0, len(arr) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            right = mid - 1
        else:
            left = mid + 1
    
    return -1

# Example usage
arr_np = np.array([2, 3, 4, 10, 40, 50, 60, 70])
target = 40
result = binary_search_numpy(arr_np, target)
print(f"Custom NumPy Binary Search: Element {target} found at index {result}")  # Output: Element 40 found at index 4
```

## 5. Practical Examples

### Example 1: Finding First and Last Occurrence

Binary search can be modified to find the first or last occurrence of a target in a sorted array with duplicates.

```python
def find_first_last(arr, target):
    def find_first(arr, target):
        left, right = 0, len(arr) - 1
        first = -1
        while left <= right:
            mid = (left + right) // 2
            if arr[mid] == target:
                first = mid
                right = mid - 1  # Continue searching left
            elif arr[mid] > target:
                right = mid - 1
            else:
                left = mid + 1
        return first
    
    def find_last(arr, target):
        left, right = 0, len(arr) - 1
        last = -1
        while left <= right:
            mid = (left + right) // 2
            if arr[mid] == target:
                last = mid
                left = mid + 1  # Continue searching right
            elif arr[mid] > target:
                right = mid - 1
            else:
                left = mid + 1
        return last
    
    return find_first(arr, target), find_last(arr, target)

# Example usage
arr = [2, 3, 3, 3, 4, 10, 10, 40]
target = 3
first, last = find_first_last(arr, target)
print(f"First occurrence of {target} at index {first}, last at index {last}")  # Output: First occurrence of 3 at index 1, last at index 3
```

### Example 2: Searching in a Rotated Sorted Array

Binary search can be adapted for a sorted array that has been rotated.

```python
def search_rotated_array(arr, target):
    left, right = 0, len(arr) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if arr[mid] == target:
            return mid
        
        # Check if left half is sorted
        if arr[left] <= arr[mid]:
            # Check if target is in left half
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        # Right half is sorted
        else:
            # Check if target is in right half
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1
    
    return -1

# Example usage
arr = [4, 5, 6, 7, 0, 1, 2]
target = 0
result = search_rotated_array(arr, target)
print(f"Search in rotated array: Element {target} found at index {result}")  # Output: Element 0 found at index 4
```

## 6. Notes and Considerations

- **Sorting Requirement**: Binary search requires the array to be sorted. Use `sorted()` for lists or `np.sort()` for NumPy arrays if needed.
- **Integer Overflow**: For very large arrays, use `mid = left + (right - left) // 2` instead of `(left + right) // 2` to avoid overflow.
- **NumPy Advantage**: `np.searchsorted` is optimized for large arrays and can handle multiple targets efficiently.
- **Edge Cases**:
  - Empty array: Return -1.
  - Target not in array: Return -1.
  - Duplicates: Standard binary search returns any matching index; modify for first/last occurrence if needed.
