## 1. Introduction to Bubble Sort

Bubble sort works by repeatedly comparing and swapping adjacent elements in an array. In each pass, the largest (or smallest, depending on the order) unsorted element "bubbles up" to its correct position.

**Key Characteristics**:
- **Time Complexity**: O(n²) in worst and average cases, O(n) in best case (optimized version).
- **Space Complexity**: O(1) as it sorts in-place.
- **Stability**: Bubble sort is stable, preserving the relative order of equal elements.

**Use Cases**:
- Suitable for small datasets or educational purposes due to its simplicity.
- Not recommended for large datasets due to poor performance compared to algorithms like quicksort or mergesort.

## 2. Standard Bubble Sort

The standard implementation performs pairwise comparisons and swaps until the array is sorted.

```python
def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        # Last i elements are already in place
        for j in range(0, n - i - 1):
            # Compare adjacent elements
            if arr[j] > arr[j + 1]:
                # Swap if the element found is greater than the next element
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = bubble_sort(arr.copy())  # Use copy to preserve original array
print("Sorted array:", sorted_arr)  # Output: [11, 12, 22, 25, 34, 64, 90]
```

## 3. Optimized Bubble Sort

The optimized version stops early if no swaps are needed in a pass, indicating the array is already sorted.

```python
def bubble_sort_optimized(arr):
    n = len(arr)
    for i in range(n):
        swapped = False  # Flag to track swaps
        
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        
        # If no swapping occurred, array is sorted
        if not swapped:
            break
    
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = bubble_sort_optimized(arr.copy())
print("Sorted array (optimized):", sorted_arr)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Test with nearly sorted array
arr_nearly_sorted = [11, 12, 22, 25, 34, 90, 64]
print("Nearly sorted array:", arr_nearly_sorted)
sorted_arr = bubble_sort_optimized(arr_nearly_sorted.copy())
print("Sorted array (optimized):", sorted_arr)  # Output: [11, 12, 22, 25, 34, 64, 90]
```

## 4. Bubble Sort with NumPy Arrays

Bubble sort can be applied to NumPy arrays, though NumPy's built-in `np.sort` is more efficient for large datasets.

### 4.2 Bubble Sort for NumPy Arrays

```python
def bubble_sort_numpy(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

# Example usage
arr_np = np.array([64, 34, 25, 12, 22, 11, 90])
print("Original NumPy array:", arr_np)
sorted_arr_np = bubble_sort_numpy(arr_np.copy())
print("Sorted NumPy array:", sorted_arr_np)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Using NumPy's built-in sort for comparison
print("Sorted with np.sort:", np.sort(arr_np))  # Output: [11, 12, 22, 25, 34, 64, 90]
```

## 5. Practical Examples

### Example 1: Sorting in Descending Order

Modify bubble sort to sort in descending order.

```python
def bubble_sort_descending(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] < arr[j + 1]:  # Change > to < for descending order
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = bubble_sort_descending(arr.copy())
print("Sorted array (descending):", sorted_arr)  # Output: [90, 64, 34, 25, 22, 12, 11]
```

### Example 2: Sorting Strings

Bubble sort can be used to sort arrays of strings.

```python
def bubble_sort_strings(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

# Example usage
arr_strings = ["banana", "apple", "cherry", "date"]
print("Original string array:", arr_strings)
sorted_strings = bubble_sort_strings(arr_strings.copy())
print("Sorted string array:", sorted_strings)  # Output: ['apple', 'banana', 'cherry', 'date']
```

### Example 3: Performance Comparison

Compare bubble sort with NumPy's built-in sort for a large array.

```python
import time
import numpy as np

# Generate a large random array
np.random.seed(42)
arr = np.random.randint(0, 1000, 1000).tolist()
arr_np = np.array(arr)

# Bubble sort performance
start_time = time.time()
bubble_sort(arr.copy())
bubble_time = time.time() - start_time
print(f"Bubble sort time: {bubble_time:.4f} seconds")

# NumPy sort performance
start_time = time.time()
np.sort(arr_np)
numpy_time = time.time() - start_time
print(f"NumPy sort time: {numpy_time:.4f} seconds")
```

## 6. Notes and Considerations

- **Optimization**: The optimized version with the `swapped` flag significantly improves performance for nearly sorted arrays.
- **NumPy Advantage**: For large datasets, use `np.sort` instead of bubble sort, as it uses a more efficient algorithm (quicksort or timsort).
- **Edge Cases**:
  - Empty array: Returns empty array.
  - Single element: Already sorted.
  - Duplicate elements: Handled correctly due to stability.
- **Practical Use**: Bubble sort is primarily educational; prefer built-in sorting functions (`sorted`, `np.sort`) for production code.
