## 1. Introduction to Insertion Sort

Insertion sort works by dividing the array into a sorted and an unsorted portion. It iteratively takes an element from the unsorted portion and inserts it into the correct position in the sorted portion.

**Key Characteristics**:
- **Time Complexity**: O(n²) in worst and average cases, O(n) in best case (nearly sorted array).
- **Space Complexity**: O(1) as it sorts in-place.
- **Stability**: Insertion sort is stable, preserving the relative order of equal elements.
- **Adaptive**: Performs efficiently on nearly sorted arrays.

**Use Cases**:
- Suitable for small datasets or nearly sorted arrays.
- Often used as a building block for more complex algorithms (e.g., hybrid sorting algorithms).
- Not recommended for large datasets due to quadratic time complexity.

## 2. Standard Insertion Sort

The standard implementation iterates through the array, inserting each element into its correct position in the sorted portion.

```python
def insertion_sort(arr):
    # Start from the second element (index 1)
    for i in range(1, len(arr)):
        key = arr[i]  # Element to be inserted
        j = i - 1     # Index of the last element in sorted portion
        
        # Move elements of arr[0..i-1] that are greater than key
        # to one position ahead of their current position
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key  # Place key in its correct position
    
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = insertion_sort(arr.copy())  # Use copy to preserve original array
print("Sorted array:", sorted_arr)  # Output: [11, 12, 22, 25, 34, 64, 90]
```

## 3. Insertion Sort with NumPy Arrays

Insertion sort can be applied to NumPy arrays, though NumPy's built-in `np.sort` is more efficient for large datasets.

### 3.2 Insertion Sort for NumPy Arrays

```python
def insertion_sort_numpy(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Example usage
arr_np = np.array([64, 34, 25, 12, 22, 11, 90])
print("Original NumPy array:", arr_np)
sorted_arr_np = insertion_sort_numpy(arr_np.copy())
print("Sorted NumPy array:", sorted_arr_np)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Using NumPy's built-in sort for comparison
print("Sorted with np.sort:", np.sort(arr_np))  # Output: [11, 12, 22, 25, 34, 64, 90]
```

## 4. Practical Examples

### Example 1: Sorting in Descending Order

Modify insertion sort to sort in descending order.

```python
def insertion_sort_descending(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] < key:  # Change > to < for descending order
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = insertion_sort_descending(arr.copy())
print("Sorted array (descending):", sorted_arr)  # Output: [90, 64, 34, 25, 22, 12, 11]
```

### Example 2: Sorting Strings

Insertion sort can be used to sort arrays of strings.

```python
def insertion_sort_strings(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Example usage
arr_strings = ["banana", "apple", "cherry", "date"]
print("Original string array:", arr_strings)
sorted_strings = insertion_sort_strings(arr_strings.copy())
print("Sorted string array:", sorted_strings)  # Output: ['apple', 'banana', 'cherry', 'date']
```

### Example 3: Performance Comparison

Compare insertion sort with NumPy's built-in sort for a large array.

```python
import time
import numpy as np

# Generate a large random array
np.random.seed(42)
arr = np.random.randint(0, 1000, 1000).tolist()
arr_np = np.array(arr)

# Insertion sort performance
start_time = time.time()
insertion_sort(arr.copy())
insertion_time = time.time() - start_time
print(f"Insertion sort time: {insertion_time:.4f} seconds")

# NumPy sort performance
start_time = time.time()
np.sort(arr_np)
numpy_time = time.time() - start_time
print(f"NumPy sort time: {numpy_time:.4f} seconds")
```

## 5. Notes and Considerations

- **Efficiency**: Insertion sort is efficient for small or nearly sorted arrays due to its adaptive nature.
- **NumPy Advantage**: For large datasets, use `np.sort` instead of insertion sort, as it uses a more efficient algorithm (quicksort or timsort).
- **Edge Cases**:
  - Empty array: Returns empty array.
  - Single element: Already sorted.
  - Duplicate elements: Handled correctly due to stability.
- **Practical Use**: Insertion sort is useful for educational purposes or when dealing with small or partially sorted datasets. For production code, prefer built-in sorting functions (`sorted`, `np.sort`).
