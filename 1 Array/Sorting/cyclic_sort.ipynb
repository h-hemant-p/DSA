# Cyclic Sort Implementation in Python

Cyclic Sort is an in-place, unstable sorting algorithm that is optimal for minimizing the number of writes to the array. It is particularly useful when the array contains distinct integers from 1 to n. The algorithm works by placing each element in its correct position based on its value (i.e., value `k` should be at index `k-1`).

```python
def cyclic_sort(arr):
    """Sorts an array of distinct integers from 1 to n using Cyclic Sort.
    Args:
        arr (list): Input array of integers.
    Returns:
        None: The array is sorted in-place.
    """
    i = 0
    while i < len(arr):
        correct_index = arr[i] - 1  # Correct index for value arr[i]
        if arr[i] != arr[correct_index]:
            arr[i], arr[correct_index] = arr[correct_index], arr[i]  # Swap
        else:
            i += 1

# Function to print the array
def print_array(arr):
    print("Array:", arr)
```

## Example Usage

Let's test the Cyclic Sort algorithm with an example array.

```python
# Test the Cyclic Sort implementation
arr = [3, 5, 2, 1, 4]
print("Before sorting:")
print_array(arr)
cyclic_sort(arr)
print("After sorting:")
print_array(arr)
```

## Explanation

- **Input Array**: `[3, 5, 2, 1, 4]`
- **Expected Output**: `[1, 2, 3, 4, 5]`
- The algorithm iterates through the array. For each element, it checks if it is in its correct position (value `k` should be at index `k-1`).
- If an element is not in its correct position, it swaps it with the element at its correct index.
- If an element is already in its correct position, the index is incremented.
- This process continues until all elements are in their correct positions.

## Time and Space Complexity
- **Time Complexity**: O(n) in the worst case, where n is the length of the array. The while loop can iterate up to 2n-1 times in the worst case.
- **Space Complexity**: O(1) as it is an in-place algorithm.

This implementation assumes the input array contains distinct integers from 1 to n. If the array contains other values, additional validation would be needed.

```python
# Test Case 1: Already sorted array
arr1 = [1, 2, 3, 4, 5]
print("Test Case 1 - Already sorted array:")
print("Before sorting:")
print_array(arr1)
cyclic_sort(arr1)
print("After sorting:")
print_array(arr1)

# Test Case 2: Reverse sorted array
arr2 = [5, 4, 3, 2, 1]
print("\nTest Case 2 - Reverse sorted array:")
print("Before sorting:")
print_array(arr2)
cyclic_sort(arr2)
print("After sorting:")
print_array(arr2)

# Test Case 3: Array with random order
arr3 = [2, 6, 4, 3, 1, 5]
print("\nTest Case 3 - Random order array:")
print("Before sorting:")
print_array(arr3)
cyclic_sort(arr3)
print("After sorting:")
print_array(arr3)
```
