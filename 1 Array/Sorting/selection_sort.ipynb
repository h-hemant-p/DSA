## 1. Introduction to Selection Sort

Selection sort divides the array into a sorted and an unsorted portion. In each iteration, it finds the minimum element in the unsorted portion and swaps it with the first element of the unsorted portion, expanding the sorted portion.

**Key Characteristics**:
- **Time Complexity**: O(n²) in all cases (worst, average, and best).
- **Space Complexity**: O(1) as it sorts in-place.
- **Stability**: Selection sort is not stable, as it may change the relative order of equal elements.
- **Non-Adaptive**: Performs the same number of comparisons regardless of the input's initial order.

**Use Cases**:
- Suitable for small datasets or when memory is limited (in-place sorting).
- Useful when the cost of swapping is low or when auxiliary memory is constrained.
- Not recommended for large datasets due to quadratic time complexity.

## 2. Standard Selection Sort

The standard implementation iterates through the array, finding the minimum element in the unsorted portion and swapping it with the first element of that portion. This version includes input validation and edge case handling.

```python
def selection_sort(arr):
    # Input validation
    if not isinstance(arr, list):
        raise TypeError("Input must be a list")
    if not arr:
        return arr  # Empty array
    if len(arr) == 1:
        return arr  # Single element is already sorted
    
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            try:
                if arr[j] < arr[min_idx]:
                    min_idx = j
            except TypeError as e:
                raise TypeError("Elements must be comparable") from e
        
        # Swap only if a different minimum is found
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
    
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = selection_sort(arr.copy())  # Use copy to preserve original array
print("Sorted array:", sorted_arr)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Edge cases
print("Empty array:", selection_sort([]))  # Output: []
print("Single element:", selection_sort([42]))  # Output: [42]

# Test invalid input
try:
    selection_sort([1, 2, "three"])  # Should raise TypeError
except TypeError as e:
    print("Error:", str(e))
```

## 3. Selection Sort with NumPy Arrays

Selection sort can be applied to NumPy arrays, though NumPy's built-in `np.sort` is more efficient for large datasets. This version includes input validation and optimized swapping.

### 3.2 Selection Sort for NumPy Arrays

```python
def selection_sort_numpy(arr):
    # Input validation
    if not isinstance(arr, np.ndarray):
        raise TypeError("Input must be a NumPy array")
    if arr.size == 0:
        return arr  # Empty array
    if arr.size == 1:
        return arr  # Single element is already sorted
    
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            try:
                if arr[j] < arr[min_idx]:
                    min_idx = j
            except TypeError as e:
                raise TypeError("Elements must be comparable") from e
        
        # Swap using NumPy indexing
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]  # Works with NumPy arrays
    
    return arr

# Example usage
arr_np = np.array([64, 34, 25, 12, 22, 11, 90])
print("Original NumPy array:", arr_np)
sorted_arr_np = selection_sort_numpy(arr_np.copy())
print("Sorted NumPy array:", sorted_arr_np)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Using NumPy's built-in sort for comparison
print("Sorted with np.sort:", np.sort(arr_np))  # Output: [11, 12, 22, 25, 34, 64, 90]

# Edge cases
print("Empty NumPy array:", selection_sort_numpy(np.array([])))  # Output: []
print("Single element:", selection_sort_numpy(np.array([42])))  # Output: [42]
```

## 4. Practical Examples

### Example 1: Sorting in Descending Order

Modify selection sort to sort in descending order.

```python
def selection_sort_descending(arr):
    # Input validation
    if not isinstance(arr, list):
        raise TypeError("Input must be a list")
    if not arr:
        return arr
    if len(arr) == 1:
        return arr
    
    n = len(arr)
    for i in range(n):
        max_idx = i
        for j in range(i + 1, n):
            try:
                if arr[j] > arr[max_idx]:  # Change < to > for descending order
                    max_idx = j
            except TypeError as e:
                raise TypeError("Elements must be comparable") from e
        if max_idx != i:
            arr[i], arr[max_idx] = arr[max_idx], arr[i]
    return arr

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = selection_sort_descending(arr.copy())
print("Sorted array (descending):", sorted_arr)  # Output: [90, 64, 34, 25, 22, 12, 11]
```

### Example 2: Sorting Strings

Selection sort can be used to sort arrays of strings.

```python
def selection_sort_strings(arr):
    # Input validation
    if not isinstance(arr, list):
        raise TypeError("Input must be a list")
    if not arr:
        return arr
    if len(arr) == 1:
        return arr
    
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            try:
                if arr[j] < arr[min_idx]:
                    min_idx = j
            except TypeError as e:
                raise TypeError("Elements must be comparable") from e
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# Example usage
arr_strings = ["banana", "apple", "cherry", "date"]
print("Original string array:", arr_strings)
sorted_strings = selection_sort_strings(arr_strings.copy())
print("Sorted string array:", sorted_strings)  # Output: ['apple', 'banana', 'cherry', 'date']
```

### Example 3: Performance Comparison

Compare selection sort with NumPy's built-in sort for a large array, using multiple runs for reliable timing.

```python
import time
import numpy as np

# Generate a large random array
np.random.seed(42)
arr = np.random.randint(0, 1000, 1000).tolist()
arr_np = np.array(arr)

# Selection sort performance (average of 3 runs)
selection_times = []
for _ in range(3):
    start_time = time.time()
    selection_sort(arr.copy())
    selection_times.append(time.time() - start_time)
print(f"Selection sort average time: {sum(selection_times) / len(selection_times):.4f} seconds")

# NumPy sort performance (average of 3 runs)
numpy_times = []
for _ in range(3):
    start_time = time.time()
    np.sort(arr_np.copy())
    numpy_times.append(time.time() - start_time)
print(f"NumPy sort average time: {sum(numpy_times) / len(numpy_times):.4f} seconds")
```

## 5. Notes and Considerations

- **Non-Adaptive**: Selection sort performs the same number of comparisons regardless of the input's order, unlike insertion sort.
- **Non-Stable**: Selection sort does not preserve the relative order of equal elements (e.g., in `[5a, 5b]`, `5a` and `5b` may swap positions).
- **NumPy Advantage**: For large datasets, use `np.sort` instead of selection sort, as it uses a more efficient algorithm (quicksort or timsort).
- **Edge Cases**:
  - Empty array: Handled explicitly, returns empty array.
  - Single element: Handled explicitly, already sorted.
  - Non-comparable elements: Raises TypeError with a clear message.
- **Practical Use**: Selection sort is primarily educational or used when minimizing swaps is important. For production code, prefer built-in sorting functions (`sorted`, `np.sort`).
