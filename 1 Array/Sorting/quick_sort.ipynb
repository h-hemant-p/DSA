## 1. Introduction to Quick Sort

Quicksort works by selecting a **pivot** element, partitioning the array such that elements smaller than the pivot are on the left and larger elements are on the right, and then recursively sorting the subarrays. The choice of pivot affects performance.

**Key Characteristics**:
- **Time Complexity**: O(n log n) on average, O(n²) in the worst case (e.g., already sorted array with a bad pivot choice).
- **Space Complexity**: O(log n) for the recursive call stack (in-place implementation).
- **Stability**: Quicksort is not stable, as it may change the relative order of equal elements.
- **Adaptive**: Performs better on random or partially sorted arrays with a good pivot selection.

**Use Cases**:
- Suitable for large datasets due to its average-case efficiency.
- Widely used in general-purpose sorting (e.g., Python's `sorted` function uses a hybrid of quicksort and other algorithms).
- Less efficient for small arrays or when stability is required.

## 2. Standard Quick Sort (Recursive, In-Place)

The standard implementation uses an in-place partitioning scheme with the last element as the pivot.

```python
def quick_sort(arr, low, high):
    def partition(arr, low, high):
        pivot = arr[high]  # Choose the last element as pivot
        i = low - 1  # Index of smaller element
        
        for j in range(low, high):
            # If current element is smaller than or equal to pivot
            if arr[j] <= pivot:
                i += 1  # Increment index of smaller element
                arr[i], arr[j] = arr[j], arr[i]
        
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1  # Return the partition index
    
    if low < high:
        # Find pivot position such that smaller elements are on the left
        pi = partition(arr, low, high)
        
        # Recursively sort elements before and after partition
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)
    
    return arr

# Wrapper function for easier usage
def quick_sort_wrapper(arr):
    # Input validation
    if not isinstance(arr, list):
        raise TypeError("Input must be a list")
    if not arr:
        return arr  # Empty array
    if len(arr) == 1:
        return arr  # Single element is already sorted
    try:
        return quick_sort(arr.copy(), 0, len(arr) - 1)
    except TypeError as e:
        raise TypeError("Elements must be comparable") from e

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = quick_sort_wrapper(arr)
print("Sorted array:", sorted_arr)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Edge cases
print("Empty array:", quick_sort_wrapper([]))  # Output: []
print("Single element:", quick_sort_wrapper([42]))  # Output: [42]
```

## 3. Quick Sort with NumPy Arrays

Quicksort can be applied to NumPy arrays, though NumPy's built-in `np.sort` is more efficient for large datasets.

### 3.2 Quick Sort for NumPy Arrays

```python
def quick_sort_numpy(arr, low, high):
    def partition(arr, low, high):
        pivot = arr[high]
        i = low - 1
        
        for j in range(low, high):
            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
        
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1
    
    if low < high:
        pi = partition(arr, low, high)
        quick_sort_numpy(arr, low, pi - 1)
        quick_sort_numpy(arr, pi + 1, high)
    
    return arr

# Wrapper function for NumPy arrays
def quick_sort_numpy_wrapper(arr):
    if not isinstance(arr, np.ndarray):
        raise TypeError("Input must be a NumPy array")
    if arr.size == 0:
        return arr
    if arr.size == 1:
        return arr
    try:
        return quick_sort_numpy(arr.copy(), 0, len(arr) - 1)
    except TypeError as e:
        raise TypeError("Elements must be comparable") from e

# Example usage
arr_np = np.array([64, 34, 25, 12, 22, 11, 90])
print("Original NumPy array:", arr_np)
sorted_arr_np = quick_sort_numpy_wrapper(arr_np)
print("Sorted NumPy array:", sorted_arr_np)  # Output: [11, 12, 22, 25, 34, 64, 90]

# Using NumPy's built-in sort for comparison
print("Sorted with np.sort:", np.sort(arr_np))  # Output: [11, 12, 22, 25, 34, 64, 90]

# Edge cases
print("Empty NumPy array:", quick_sort_numpy_wrapper(np.array([])))  # Output: []
print("Single element:", quick_sort_numpy_wrapper(np.array([42])))  # Output: [42]
```

## 4. Practical Examples

### Example 1: Sorting in Descending Order

Modify quicksort to sort in descending order.

```python
def quick_sort_descending(arr, low, high):
    def partition(arr, low, high):
        pivot = arr[high]
        i = low - 1
        
        for j in range(low, high):
            if arr[j] >= pivot:  # Change <= to >= for descending order
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
        
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1
    
    if low < high:
        pi = partition(arr, low, high)
        quick_sort_descending(arr, low, pi - 1)
        quick_sort_descending(arr, pi + 1, high)
    
    return arr

# Wrapper function
def quick_sort_descending_wrapper(arr):
    if not isinstance(arr, list):
        raise TypeError("Input must be a list")
    if not arr:
        return arr
    if len(arr) == 1:
        return arr
    try:
        return quick_sort_descending(arr.copy(), 0, len(arr) - 1)
    except TypeError as e:
        raise TypeError("Elements must be comparable") from e

# Example usage
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = quick_sort_descending_wrapper(arr)
print("Sorted array (descending):", sorted_arr)  # Output: [90, 64, 34, 25, 22, 12, 11]
```

### Example 2: Sorting Strings

Quicksort can be used to sort arrays of strings.

```python
def quick_sort_strings(arr):
    return quick_sort_wrapper(arr)

# Example usage
arr_strings = ["banana", "apple", "cherry", "date"]
print("Original string array:", arr_strings)
sorted_strings = quick_sort_strings(arr_strings)
print("Sorted string array:", sorted_strings)  # Output: ['apple', 'banana', 'cherry', 'date']
```

### Example 3: Performance Comparison

Compare quicksort with NumPy's built-in sort for a large array, using multiple runs for reliable timing.

```python
import time
import numpy as np

# Generate a large random array
np.random.seed(42)
arr = np.random.randint(0, 1000, 1000).tolist()
arr_np = np.array(arr)

# Quick sort performance (average of 3 runs)
quick_times = []
for _ in range(3):
    start_time = time.time()
    quick_sort_wrapper(arr.copy())
    quick_times.append(time.time() - start_time)
print(f"Quick sort average time: {sum(quick_times) / len(quick_times):.4f} seconds")

# NumPy sort performance (average of 3 runs)
numpy_times = []
for _ in range(3):
    start_time = time.time()
    np.sort(arr_np.copy())
    numpy_times.append(time.time() - start_time)
print(f"NumPy sort average time: {sum(numpy_times) / len(numpy_times):.4f} seconds")
```

## 5. Notes and Considerations

- **Pivot Choice**: This implementation uses the last element as the pivot for simplicity. For better performance, consider using the median-of-three or random pivot selection to avoid worst-case O(n²) behavior on sorted arrays.
- **Non-Stable**: Quicksort does not preserve the relative order of equal elements.
- **NumPy Advantage**: For large datasets, `np.sort` is optimized and typically faster than a custom quicksort implementation.
- **Edge Cases**:
  - Empty array: Handled explicitly, returns empty array.
  - Single element: Handled explicitly, already sorted.
  - Non-comparable elements: Raises TypeError with a clear message.
- **Practical Use**: Quicksort is efficient for large datasets but is primarily educational in Python, as `sorted` or `np.sort` are optimized for production use.
