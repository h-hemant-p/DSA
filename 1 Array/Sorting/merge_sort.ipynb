# Merge Sort Implementation in Python

This notebook demonstrates the Merge Sort algorithm, a divide-and-conquer sorting algorithm that recursively divides the input array into two halves, sorts them, and then merges the sorted halves to produce a sorted array.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    # Divide the array into two halves
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    
    # Recursively sort both halves
    left = merge_sort(left)
    right = merge_sort(right)
    
    # Merge the sorted halves
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    # Compare elements from both arrays and merge in sorted order
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    # Add remaining elements from left array, if any
    result.extend(left[i:])
    # Add remaining elements from right array, if any
    result.extend(right[j:])
    
    return result
```

## Example Usage

Let's test the Merge Sort implementation with an example array.

```python
# Test the merge sort function
arr = [64, 34, 25, 12, 22, 11, 90]
print("Original array:", arr)
sorted_arr = merge_sort(arr)
print("Sorted array:", sorted_arr)
```

## Explanation

- **Time Complexity**: O(n log n) for all cases, where n is the number of elements.
- **Space Complexity**: O(n) due to the auxiliary space used for merging.
- The algorithm is stable, meaning it preserves the relative order of equal elements.
- Merge Sort is efficient for large datasets and is widely used in practice.
