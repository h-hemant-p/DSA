# Doubly Circular Linked List Implementation in Python
This notebook demonstrates a doubly circular linked list implementation using Python. The doubly circular linked list is built using a Node class and a DoublyCircularLinkedList class with methods for common operations, including forward and backward traversal. The last node's next pointer connects to the head, and the head's prev pointer connects to the last node.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyCircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            new_node.next = new_node
            new_node.prev = new_node
            return
        last = self.head.prev
        new_node.next = self.head
        new_node.prev = last
        last.next = new_node
        self.head.prev = new_node

    def prepend(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            new_node.next = new_node
            new_node.prev = new_node
            return
        last = self.head.prev
        new_node.next = self.head
        new_node.prev = last
        last.next = new_node
        self.head.prev = new_node
        self.head = new_node

    def delete(self, data):
        if not self.head:
            return
        current = self.head
        while True:
            if current.data == data:
                break
            current = current.next
            if current == self.head:
                return
        if current.next == current:
            self.head = None
            return
        current.prev.next = current.next
        current.next.prev = current.prev
        if current == self.head:
            self.head = current.next

    def display_forward(self):
        if not self.head:
            return 'Empty'
        elements = []
        current = self.head
        while True:
            elements.append(str(current.data))
            current = current.next
            if current == self.head:
                break
        return ' <-> '.join(elements) + ' <-> (head)'

    def display_backward(self):
        if not self.head:
            return 'Empty'
        elements = []
        current = self.head.prev
        while True:
            elements.append(str(current.data))
            current = current.prev
            if current == self.head.prev:
                break
        return ' <-> '.join(elements) + ' <-> (last)'
```

## Example Usage
Let's create a doubly circular linked list, perform some operations, and display the results in both forward and backward directions.

```python
# Create a new doubly circular linked list
dcll = DoublyCircularLinkedList()

# Append some elements
dcll.append(1)
dcll.append(2)
dcll.append(3)
print("After appending 1, 2, 3 (forward):", dcll.display_forward())
print("After appending 1, 2, 3 (backward):", dcll.display_backward())

# Prepend an element
dcll.prepend(0)
print("After prepending 0 (forward):", dcll.display_forward())
print("After prepending 0 (backward):", dcll.display_backward())

# Delete an element
dcll.delete(2)
print("After deleting 2 (forward):", dcll.display_forward())
print("After deleting 2 (backward):", dcll.display_backward())
```

## Explanation
- **Node Class**: Represents each element with `data`, a `next` pointer, and a `prev` pointer.
- **DoublyCircularLinkedList Class**:
  - `append`: Adds a node at the end of the list, connecting the last node to the head and the head's prev to the last node.
  - `prepend`: Adds a node at the beginning of the list, updating the head and maintaining circular connections.
  - `delete`: Removes the first occurrence of a node with the specified data, updating the circular structure.
  - `display_forward`: Returns a string representation of the list from head onward, indicating the circular nature.
  - `display_backward`: Returns a string representation of the list from the last node backward.

The doubly circular linked list uses a Python list internally only for displaying the elements, maintaining the linked structure for all operations.
