# Doubly Linked List Implementation in Python
This notebook demonstrates a doubly linked list implementation using Python. The doubly linked list is built using a Node class and a DoublyLinkedList class with methods for common operations, including forward and backward traversal.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            self.tail = new_node
            return
        new_node.prev = self.tail
        self.tail.next = new_node
        self.tail = new_node

    def prepend(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            self.tail = new_node
            return
        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node

    def delete(self, data):
        current = self.head
        while current and current.data != data:
            current = current.next
        if not current:
            return
        if current == self.head:
            self.head = current.next
            if self.head:
                self.head.prev = None
            else:
                self.tail = None
            return
        if current == self.tail:
            self.tail = current.prev
            self.tail.next = None
            return
        current.prev.next = current.next
        current.next.prev = current.prev

    def display_forward(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        return ' <-> '.join(elements)

    def display_backward(self):
        elements = []
        current = self.tail
        while current:
            elements.append(str(current.data))
            current = current.prev
        return ' <-> '.join(elements)
```

## Example Usage
Let's create a doubly linked list, perform some operations, and display the results in both forward and backward directions.

```python
# Create a new doubly linked list
dll = DoublyLinkedList()

# Append some elements
dll.append(1)
dll.append(2)
dll.append(3)
print("After appending 1, 2, 3 (forward):", dll.display_forward())
print("After appending 1, 2, 3 (backward):", dll.display_backward())

# Prepend an element
dll.prepend(0)
print("After prepending 0 (forward):", dll.display_forward())
print("After prepending 0 (backward):", dll.display_backward())

# Delete an element
dll.delete(2)
print("After deleting 2 (forward):", dll.display_forward())
print("After deleting 2 (backward):", dll.display_backward())
```

## Explanation
- **Node Class**: Represents each element with `data`, a `next` pointer, and a `prev` pointer.
- **DoublyLinkedList Class**:
  - `append`: Adds a node at the end of the list, updating `tail`.
  - `prepend`: Adds a node at the beginning of the list, updating `head`.
  - `delete`: Removes the first occurrence of a node with the specified data, handling head, tail, and middle cases.
  - `display_forward`: Returns a string representation of the list from head to tail.
  - `display_backward`: Returns a string representation of the list from tail to head.

The doubly linked list uses a Python list internally only for displaying the elements, maintaining the linked structure for all operations.
