## Example Usage
Let's create a circular linked list, perform some operations, and display the results.

```python
# Create a new circular linked list
cll = CircularLinkedList()

# Append some elements
cll.append(1)
cll.append(2)
cll.append(3)
print("After appending 1, 2, 3:", cll.display())

# Prepend an element
cll.prepend(0)
print("After prepending 0:", cll.display())

# Delete an element
cll.delete(2)
print("After deleting 2:", cll.display())
```

## Explanation
- **Node Class**: Represents each element with `data` and a `next` pointer.
- **CircularLinkedList Class**:
  - `append`: Adds a node at the end of the list, connecting the last node to the head.
  - `prepend`: Adds a node at the beginning of the list, updating the head and connecting the last node.
  - `delete`: Removes the first occurrence of a node with the specified data, maintaining the circular structure.
  - `display`: Returns a string representation of the list, indicating the circular nature by looping back to the head.

The circular linked list uses a Python list internally only for displaying the elements, maintaining the linked structure for all operations.
