# Python String Methods Notes

This notebook provides a comprehensive guide to Python's built-in string methods, using the example string `string1 = "Hello World"`. Each method is explained with an example to demonstrate its usage.

## Initial Code Example
Let's start with the basic code:

```python
string1 = "Hello World"
print(string1)      # Output: Hello World
print(string1[0])   # Output: H
```

## Explanation of Initial Code
- `string1 = "Hello World"`: Creates a string variable with the value `"Hello World"`.
- `print(string1)`: Prints the entire string: `Hello World`.
- `print(string1[0])`: Prints the character at index 0, which is `H` (strings are zero-indexed in Python).

## String Methods Overview
Python strings are immutable sequences of characters and come with a variety of built-in methods. Below, we group these methods by functionality and provide examples using `string1`.

## 1. Case Conversion Methods
These methods modify the case of the string.

```python
# capitalize(): Capitalizes the first character, lowercases the rest
print(string1.capitalize())  # Output: Hello world

# casefold(): Converts to lowercase for case-insensitive comparisons
print(string1.casefold())  # Output: hello world

# lower(): Converts all characters to lowercase
print(string1.lower())  # Output: hello world

# upper(): Converts all characters to uppercase
print(string1.upper())  # Output: HELLO WORLD

# title(): Capitalizes the first letter of each word
print(string1.title())  # Output: Hello World

# swapcase(): Swaps case of each character
print(string1.swapcase())  # Output: hELLO wORLD
```

## 2. Searching and Finding Methods
These methods help locate substrings or characters.

```python
# count(sub): Counts occurrences of sub
print(string1.count("l"))  # Output: 3

# find(sub): Returns lowest index of sub or -1 if not found
print(string1.find("World"))  # Output: 6

# rfind(sub): Returns highest index of sub or -1 if not found
print(string1.rfind("l"))  # Output: 9

# index(sub): Like find(), but raises ValueError if not found
print(string1.index("o"))  # Output: 4

# rindex(sub): Like rfind(), but raises ValueError if not found
print(string1.rindex("l"))  # Output: 9

# startswith(prefix): Checks if string starts with prefix
print(string1.startswith("He"))  # Output: True

# endswith(suffix): Checks if string ends with suffix
print(string1.endswith("ld"))  # Output: True
```

## 3. String Testing Methods
These methods check the content or properties of the string.

```python
# isalnum(): Checks if all characters are alphanumeric
print(string1.isalnum())  # Output: False (contains space)

# isalpha(): Checks if all characters are alphabetic
print(string1.isalpha())  # Output: False (contains space)

# isascii(): Checks if all characters are ASCII
print(string1.isascii())  # Output: True

# isdigit(): Checks if all characters are digits
print(string1.isdigit())  # Output: False

# isdecimal(): Checks if all characters are decimal digits
print(string1.isdecimal())  # Output: False

# isnumeric(): Checks if all characters are numeric
print(string1.isnumeric())  # Output: False

# islower(): Checks if all cased characters are lowercase
print(string1.islower())  # Output: False

# isupper(): Checks if all cased characters are uppercase
print(string1.isupper())  # Output: False

# istitle(): Checks if string is title-cased
print(string1.istitle())  # Output: True

# isspace(): Checks if all characters are whitespace
print(string1.isspace())  # Output: False

# isprintable(): Checks if all characters are printable
print(string1.isprintable())  # Output: True

# isidentifier(): Checks if string is a valid identifier
print("hello_world".isidentifier())  # Output: True
```

## 4. String Modification Methods
These methods return a new string with modifications.

```python
# replace(old, new[, count]): Replaces old with new
print(string1.replace("World", "Python"))  # Output: Hello Python

# strip([chars]): Removes leading/trailing chars (default: whitespace)
print("  Hello World  ".strip())  # Output: Hello World

# lstrip([chars]): Removes leading chars
print("  Hello World".lstrip())  # Output: Hello World

# rstrip([chars]): Removes trailing chars
print("Hello World  ".rstrip())  # Output: Hello World

# zfill(width): Pads string with zeros on the left to reach width
print(string1.zfill(15))  # Output: 0000Hello World

# center(width[, fillchar]): Centers string in width
print(string1.center(15, "*"))  # Output: **Hello World***

# ljust(width[, fillchar]): Left-justifies string in width
print(string1.ljust(15, "*"))  # Output: Hello World****

# rjust(width[, fillchar]): Right-justifies string in width
print(string1.rjust(15, "*"))  # Output: ****Hello World
```

## 5. Splitting and Joining Methods
These methods split strings or join iterables into strings.

```python
# split(sep=None, maxsplit=-1): Splits string into a list
print(string1.split())  # Output: ['Hello', 'World']

# rsplit(sep=None, maxsplit=-1): Splits from the right
print(string1.rsplit("o"))  # Output: ['Hell', ' W', 'rld']

# splitlines([keepends]): Splits on line breaks
print("Hello\nWorld".splitlines())  # Output: ['Hello', 'World']

# join(iterable): Joins elements of iterable with string
print("-".join(["Hello", "World"]))  # Output: Hello-World

# partition(sep): Splits into tuple (before, sep, after)
print(string1.partition(" "))  # Output: ('Hello', ' ', 'World')

# rpartition(sep): Splits from the right into tuple
print(string1.rpartition(" "))  # Output: ('Hello', ' ', 'World')
```

## 6. Encoding and Decoding Methods
These methods handle string encoding and formatting.

```python
# encode(encoding='utf-8', errors='strict'): Encodes string to bytes
print(string1.encode())  # Output: b'Hello World'

# expandtabs(tabsize=8): Replaces tabs with spaces
print("Hello\tWorld".expandtabs(4))  # Output: Hello    World

# format(*args, **kwargs): Formats string using placeholders
print("Hello {0}".format("World"))  # Output: Hello World

# format_map(mapping): Formats string using a mapping
print("Hello {name}".format_map({'name': 'World'}))  # Output: Hello World
```

## 7. Translation Methods
These methods handle character translations.

```python
# maketrans(x[, y[, z]]): Creates a translation table
trans = str.maketrans("lo", "12")
print(string1.translate(trans))  # Output: He112 W0r1d

# translate(table): Applies translation table to string
print(string1.translate(str.maketrans("", "", "l")))  # Output: Heo Word
```

## Notes
- Strings are **immutable**; all methods return a new string and do not modify the original.
- Methods like `find()`, `index()`, `count()`, etc., accept optional `start` and `end` indices to limit the search range.
- For formatting, consider using f-strings (Python 3.6+) for simpler syntax, e.g., `f"Hello {variable}"`.
