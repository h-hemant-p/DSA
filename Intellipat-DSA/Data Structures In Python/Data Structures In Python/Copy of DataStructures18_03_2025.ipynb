```python
n = 0                  #  O(1) Constant time complexity.
while n < 5:    # This loop is running 5 times.   O(5).
  print(n)
  n+=1
```

```python
N = 150                         #O(1)
for x in range(N):              #O(N)  Linear Time Complexity
  print(x)

# Final Time Complexity O(1) + O(N)
             # =         O(N+1)
```

**Whenever we are analysing the worst case time complexity we ignore the constants/ smaller terms
**

```python
# O(N^2 + N)
# => O(N^2)
```

```python
# Use for loop and print all the elements of a list.

li = [10,20,30,40,50]
# This syntax not allowed: for x in li: print(x)
```

```python
# We need to run a for/while loop till the length of our list.
length_li = len(li)
print(length_li)
for x in range(length_li):
  print(li[x])            # Time complexity = O(N) N = Number of data elements.
```

```python
N = 5
for x in range(N):  #O(N)
  for j in range(N):   #O(N)
    print(j)
  print("Loop is over",x)

# Final Time complexity = O(N*N) => O(N^2)   Quadratic Time Complexity
```

```python
count = 0
for x in range(N):
  for j in range(N):
    for k in range(N):
      print(x,j,k)
      count+=1

print(count)   # Time complexity = O(N^3)
```

```python
N=50
for x in range(N):      # O(N)
  for j in range(1,N,3):   # O(Log N)
    print(j)

# Time complexity = O(N*logN) = O(N Log N)  Linearithmic Time Complexity.
```

```python
def fun(N):
  sum = 0
  for x in range(N):    #Outer Loop -> O(N)
    for j in range(N):   # Inner Loop -> O(N)
      sum = sum + 10    # Arithmetic Operation ->  O(1)
  return sum

# O(N^2)
```

```python
def fun2(N):   # Time complexity = O(1) Constant time complexity.
  return N*N
```

```python
i=1
N = 10
while i<N:    # O(Log N)
  print(i)
  i*=2
```

```python
i=1
N = 10
while i<N:     # O(LogN)
  for j in range(N):   # O(N)
    print(j)
  i*=2

# O(Log N * N)
# O(N log N)
```

# Arrays

```python
arr = [20,30,40,10]
# Traversal = Visiting each and every element of your data strucutre one by one.
```

```python
N = len(arr)
for x in range(N):
  print(arr[x])

# Time complexity = O(N)
# To traverse your array the time complexity is O(N)
```

```python
# To delete an element from your array the time complexity will be O(N)
```

```python
# We have a data structure that can do the same task (Deleting the first element) in constant time compexity O(1)
```

```python
# Linked Lists
```

```python
# O(N+N)
# O(2N)
# Final Answer will be O(N)
# O(N(Log N) + N)
# Final Complexity = O(N log N)
```

```python
n=10
for i in range(n):   # O(Log N)
  for j in range(n):  #(O Log N)
    print(j/2)
  print(i/2)

# Final = O(LogN*LogN)  =  O(LogN)^2       Quadratic Logarithmic Time Complexity.
```
