```python
class Node:
  def __init__(self,value,next=None,prev=None):
    self.value=value
    self.next=next
    self.prev = prev


class DLL:
  def __init__(self):
    self.head = None

  def insertBeg(self,value):   #[7,4,2,1,6]
    new_node = Node(value,self.head)
    if self.head==None:   # This means the linked list is empty.
      self.head = new_node
      return

    self.head.prev = new_node
    self.head = new_node

  def insertEnd(self,value):
    if self.head==None:
      self.insertBeg(value)
      return

    temp = self.head
    while temp.next!=None:
      temp = temp.next

    new_node = Node(value)
    temp.next = new_node
    new_node.prev = temp



  def display(self):
      temp = self.head
      while temp!=None:
        print(temp.value,end='-->')
        temp = temp.next
      print('End')


  def displayReverse(self):
    temp = self.head
    while temp.next!=None:  # When this loop will be over temp variable will be at last node
      temp = temp.next

    while temp!=None:
      print(temp.value,end='<-->')
      temp = temp.prev
    print('None')
```

Approach to display all the elements in your doubly linked list.
create a temp variable = head
while temp is not equal to none print the element and move temp to next node (temp=temp.next)

```python
ob = DLL()
li = [7,4,2,1,6]
ob.insertEnd(20)
for x in range(len(li)):
  ob.insertBeg(li[x])

ob.insertEnd(5)
ob.insertEnd(43)
ob.insertEnd(72)
ob.insertEnd(81)

ob.display()
ob.displayReverse()
```
