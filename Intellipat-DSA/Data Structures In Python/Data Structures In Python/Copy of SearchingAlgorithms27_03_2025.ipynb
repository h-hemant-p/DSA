# Linear Search

```python
def Linear_Search(arr,target):
  if len(arr)==0:
      return "Empty Array"
  for x in range(len(arr)):
    if target == arr[x]:
      return "Element found at: ",x

  return -1

# Time complexity of Linear Search => O(N)   N -> Number of data elements present inside our array.
```

```python
Linear_Search([],50)
```

```python
arr = [1,2,3,4,5,6,7]
target = 5
for x in range(len(arr)):
  if arr[x]==target:
    print("Element found at index:",x)
```
