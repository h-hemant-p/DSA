```python
class BST:
  def __init__(self,value):  # self = 7
    self.value=value
    self.left = None
    self.right = None

  def add_value(self,value):  #  2
    if value <= self.value:  # 2 < 4
      if self.left==None:
        self.left = BST(value)
      else:
        self.left.add_value(value)

    elif value > self.value:
      if self.right==None:
        self.right = BST(value)
      else:
        self.right.add_value(value)


```

```python
ob = BST(7)  # While creating the object we will pass the value of root node.
ob.add_value(4)
ob.add_value(9)
ob.add_value(2)
ob.add_value(6)
ob.add_value(8)
ob.add_value(10)
```
